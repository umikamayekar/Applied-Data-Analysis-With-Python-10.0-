## 1. What is Pandas?

Pandas is a Python library used to **organise, clean, manipulate and analyse data**.

```python
import pandas as pd
```

`pd` is the conventional alias for Pandas.

## 2. Series vs DataFrame

| Series | DataFrame |
|---|---|
| One-dimensional | Two-dimensional |
| One column of data | Rows and columns |
| `pd.Series([500, 900, 300])` | `pd.DataFrame({...})` |

**Memory trick:** Series = column; DataFrame = complete table.

## 3. Essential inspection commands

| Command | Purpose |
|---|---|
| `df.head()` | First 5 rows |
| `df.tail()` | Last 5 rows |
| `df.shape` | Rows and columns |
| `df.columns` | Column names |
| `df.dtypes` | Data types |
| `df.info()` | Dataset structure |
| `df.describe()` | Numeric statistics |
| `df.isnull().sum()` | Missing-value count |

> **Interview tip:** `df.shape` has no parentheses.

## 4. Selecting and filtering

```python
df["Salary"]                 # One column → Series
df[["Name", "Salary"]]       # Multiple columns → DataFrame
df[df["Salary"] > 25000]     # Filter rows
```

## 5. `.loc` vs `.iloc`

```python
df.loc[0, "Name"]  # Select by labels
df.iloc[0, 0]      # Select by integer positions
```

- `.loc` → labels
- `.iloc` → positions
