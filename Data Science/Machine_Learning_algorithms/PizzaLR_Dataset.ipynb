```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

#load dataset
df = pd.read_excel("pizza_price_dataset.xlsx")
df.head()

```

```python
X = df[['Diameter_Inches']]
Y = df['Price_USD']

model = LinearRegression()
model.fit(X, Y)
```

```python
#find intercept and slope
m = model.coef_[0]
b = model.intercept_
print(f"Slope m = {m}")
print(f"Intercept b = {b}")
```

```python
#prediction
new_pizza = [[25]]
prediction = model.predict(new_pizza)
print(f"new 25 inche pizza price = {prediction[0]:.2f}")
```

```python
plt.scatter(X,Y,color='green',label='Data Point')
plt.plot(X, model.predict(X), color='red',label="Regression line")
plt.xlabel("inches")
plt.ylabel("Price $")
plt.title("Pizza price prediction")
plt.legend()
plt.show()
```
