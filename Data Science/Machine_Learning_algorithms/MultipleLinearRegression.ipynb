```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

#step 1 Data set
data = {
    "size":[1000,1500,2000,1200,1800],
    "bedrooms":[2,3,3,2,4],
    "price":[150,220,280,170,260]
}

df = pd.DataFrame(data)
df
```

```python
x = df[['size','bedrooms']]
y = df['price']

model = LinearRegression()
model.fit(x,y)
```

```python
#intercept (B0)
print("intercept (b0) = ",model.intercept_)
#coefficients (b1,b2)
print("coefficients (b1, b2): ",model.coef_)
```

```python
# step predict for new house 
new_house = np.array([[1600,3]])
predicted_price = model.predict(new_house)
print(f"predicted price for new house ${predicted_price[0]:.2f}")
```

```python
from sklearn.metrics import mean_squared_error, r2_score

y_pred = model.predict(x)
mse = mean_squared_error(y,y_pred)

r2 = r2_score(y, y_pred)
print("mean sqaured error", mse)
print("R- Squared", r2)
```

```python
plt.scatter(y,y_pred, color="red", label="predicted vs actual")
plt.plot([y.min(),y.max()],color="black",linestyle="--",label="perfect fit")
plt.xlabel("actual price")
plt.ylabel("predicted price")
plt.legend()
plt.show()
```
