# NumPy = Numerical Python
Think of it as:

A super-fast mathematical engine for large amounts of data.

Instead of:

sales = [1200,1500,1800]

we use:

import numpy as np

sales = np.array([1200,1500,1800])

# Why use NumPy instead of Python lists?

* Faster calculations
* Less memory usage
* Supports vectorized operations
* Foundation of Pandas, Scikit-Learn, TensorFlow, and many data science tools
